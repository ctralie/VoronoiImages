Demo the Voronoi image code

```python
%load_ext snakeviz
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
import skimage.io
import time
import torch
import glob

import sys
sys.path.append("../src")
from imgvoronoi import *
from tsp import *
```

```python
## Parameters
device = 'cpu'
n_points = 2000
n_iters = 100
I = skimage.io.imread("../data/images/layla.png")
N = min(I.shape[0], I.shape[1])
off = 70
I = I[0:N, off:N+off, ...]
plt.imshow(I)
print("Compression ratio", I.shape[0]*I.shape[1]/n_points)
```

```python
tic = time.time()
splat_intervals = [(0, 250), (0, 500), (0, 1000), (0, 2000)]
J, X, final_cost = get_voronoi_image(I, device, n_points, n_neighbs=2, temperature=20, n_iters=n_iters, splat_intervals=splat_intervals, verbose=True, plot_iter_interval=10)
print("X.shape", X.shape)
rmse = np.sqrt(final_cost/I.size)
print("Elapsed Time: {:.3f}".format(time.time()-tic))
print("Final cost {:.3f}, RMSE: {:.3f}".format(final_cost, rmse*255))
plt.figure()
plt.imshow(J)
```

```python
end = 500
#tour = get_tsp_tour(X)
tour = get_tsp_tour_piecewise(X, np.sort(np.unique(splat_intervals)))
Y = X[tour, :]
plt.figure(figsize=(15, 15))
plt.subplot(121)
plt.imshow(splat_voronoi_image_1nn(Y[0:end], 512, 512))
```

```python
for i in range(Y.shape[0]):
    I = splat_voronoi_image_1nn(Y[0:i+1], 512, 512)
    plt.imsave("TSP{}.png".format(i), I)
```
