```python
%load_ext snakeviz
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
import skimage.io
import time
import torch
import glob

import sys
sys.path.append("../src")
from imgvoronoi import *
from tsp import *
from videoutils import *
```

```python
## Parameters
device = 'cpu'
n_points = 10000
n_iters = 100
I = load_video("../data/videos/theo.avi")
I = I[:, :, 0:I.shape[2], :]
plt.imshow(I[:, :, 0, :])
print("Compression ratio", I.size/(n_points*6)) #xytrgb
print("{} kB".format(n_points*6/1024))
print("{} bpp".format(n_points*6*8/np.prod(I.shape[0:3])))
print("{:.3f} % of pixels".format(100*n_points/I.size))
```

```python
tic = time.time()
J, X, final_cost = get_voronoi_image(I, device, n_points, n_neighbs=2, temperature=20, n_iters=n_iters, verbose=True, plot_iter_interval=0)
print("X.shape", X.shape)
rmse = np.sqrt(final_cost/I.size)
print("Elapsed Time: {:.3f}".format(time.time()-tic))
print("Final cost {:.3f}, RMSE: {:.3f}".format(final_cost, rmse*255))
plt.figure()
plt.imshow(J[:, :, 0, :])
```
