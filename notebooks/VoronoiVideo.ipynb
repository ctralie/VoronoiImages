```python
%load_ext snakeviz
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
import skimage.io
import time
import torch
import glob

import sys
sys.path.append("../src")
from imgvoronoi import *
from tsp import *
from videoutils import *
```

```python
## Parameters
device = 'cuda'
I = load_video("../data/videos/theo.avi")
plt.imshow(I[:, :, 0, :])
```

```python
from skimage.transform import resize
IDown = np.zeros((I.shape[0]//4, I.shape[1]//4, I.shape[2], I.shape[3]))
for i in range(I.shape[2]):
    image_resized = resize(I[:, :, i, :], (I.shape[0]//4, I.shape[1]//4), anti_aliasing=True)
    IDown[:, :, i, :] = image_resized
```

```python
tic = time.time()
J, X, final_cost = get_voronoi_image(IDown, device, sites=30000, n_neighbs=2, temperature=20, n_iters=500, verbose=True, plot_iter_interval=50)
print("X.shape", X.shape)
rmse = np.sqrt(final_cost/I.size)
print("Elapsed Time: {:.3f}".format(time.time()-tic))
print("Final cost {:.3f}, RMSE: {:.3f}".format(final_cost, rmse*255))
plt.figure()
plt.imshow(J[:, :, 0, :])
```

```python
tic = time.time()
J, X, final_cost = get_voronoi_image(I, device, sites=X[:, 0:3], n_neighbs=2, temperature=20, n_iters=100, verbose=True, plot_iter_interval=10, lr=2e-3)
print("X.shape", X.shape)
rmse = np.sqrt(final_cost/I.size)
print("Elapsed Time: {:.3f}".format(time.time()-tic))
print("Final cost {:.3f}, RMSE: {:.3f}".format(final_cost, rmse*255))
plt.figure()
plt.imshow(J[:, :, 0, :])
```
